## Problem Statement


You are a data scientist working for a sports analytics company. The dataset, named **"sports_league_data.csv"** includes a variety of attributes for players in the league. The columns in the dataset are as follows:

- **player_id:** A unique identifier for each player.
- **age:** Age of the player.
- **position:** The position the player typically plays in (e.g., Forward, Midfielder, Defender, Goalkeeper).
- **team:** The team the player belongs to.
- **games_played:** The total number of games played by the player in the season.
- **player_rating:** An overall performance rating for the player for the season.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

**Import Necessary Libraries**

```python
df = pd.read_csv("sports_league_data.csv")
print(df.shape)
df.head(20)
```

## Task1

1. Import the data from the "sports_league_data.csv" file.
2. display the number of rows and columns. 
3. Display the first few rows of the dataset to get an overview.


```python
mean_age = round(df.age.mean(), 2)
std_age = round(df.age.std(), 2)

left_std_age_1 = round((mean_age-std_age), 2)
right_std_age_1 = round((mean_age+std_age), 2)

left_std_age_2 = round((mean_age-(2*std_age)), 2)
right_std_age_2 = round((mean_age+(2*std_age)), 2)

left_std_age_3 = round((mean_age-(3*std_age)), 2)
right_std_age_3 = round((mean_age+(3*std_age)), 2)

left_std_age_1, right_std_age_1
```

## Task2:

**Age Analysis:**
- Mean: Calculate the average age of players in the league.
- Standard Deviation Intervals:
    - Calculate the age range within one standard deviation of the mean.
    - Calculate the age range within two standard deviations of the mean.
    - Calculate the age range within three standard deviations of the mean.
  

```python
mean_rating = round(df.player_rating.mean(), 2)
std_rating = round(df.player_rating.std(), 2)

left_rating1 = round((mean_rating - std_rating), 2)
right_rating1 = round((mean_rating + std_rating), 2)

left_rating2 = round((mean_rating - (2*std_rating)), 2)
right_rating2 = round((mean_rating + (2*std_rating)), 2)

left_rating3 = round((mean_rating - (3*std_rating)), 2)
right_rating3 = round((mean_rating + (3*std_rating)), 2)

left_rating1, right_rating1
```

## Task3:

**Player Rating Analysis:**
- Mean: Determine the average player rating across the league.
- Standard Deviation Intervals:
    - Calculate the player rating range within one standard deviation of the mean.
    - Calculate the player rating range within two standard deviations of the mean.
    - Calculate the player rating range within three standard deviations of the mean.

## Task4:

Create histogram with Kernel Density Estimation (KDE) plots for both the 'age' and 'player_rating' columns from the dataset. While constructing these plots, ensure to visually represent the standard deviation intervals on the graph.


- **'age'**

```python
# AGE DISTRIBUTION
plt.figure(figsize=(10, 6))

sns.histplot(data=df, x="age", kde=True)

# Mean
plt.axvline(mean_age, color="black", linestyle="-", linewidth=2, label="Mean")

# Standard Deviations
plt.axvline(left_std_age_1, linestyle="--", color="blue", label="-1 SD")
plt.axvline(right_std_age_1, linestyle="--", color="blue", label="+1 SD")

plt.axvline(left_std_age_2, linestyle="--", color="red", label="-2 SD")
plt.axvline(right_std_age_2, linestyle="--", color="red", label="+2 SD")

plt.axvline(left_std_age_3, linestyle="--", color="orange", label="-3 SD")
plt.axvline(right_std_age_3, linestyle="--", color="orange", label="+3 SD")

plt.title("Age Distribution with KDE")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.legend()

plt.show()
```

- **'player_rating'**

```python
# PLAYER RATING DISTRIBUTION
plt.figure(figsize=(10, 6))

sns.histplot(data=df, x="player_rating", kde=True)

# Mean
plt.axvline(mean_rating, color="black", linestyle="-", linewidth=2, label="Mean")

# Standard Deviations
plt.axvline(left_rating1, linestyle="--", label="-1 SD")
plt.axvline(right_rating1, linestyle="--", label="+1 SD")

plt.axvline(left_rating2, linestyle="--", label="-2 SD")
plt.axvline(right_rating2, linestyle="--", label="+2 SD")

plt.axvline(left_rating3, linestyle="--", label="-3 SD")
plt.axvline(right_rating3, linestyle="--", label="+3 SD")

plt.title("Player Rating Distribution with KDE")
plt.xlabel("Player Rating")
plt.ylabel("Frequency")
plt.legend()

plt.show()
```

## Task5

- Calculate the lower and upper boundaries for the 'Age' and 'Player_Rating' columns. 
- For each column, identify values that fall outside the calculated lower and upper boundaries. These values are considered outliers.

- **Display age outliers**

```python
df.loc[(df.age < left_std_age_3) | (df.age > right_std_age_3)]
```

- **Display player_rating outliers**

```python
df.loc[(df.player_rating < left_rating3) | (df.player_rating > right_rating3)]
```

## Task6

-  Remove outliers from both the 'age' and 'player_rating' columns and then visualize the adjusted data using histogram plots.


```python
df_new_age = df.loc[(df.age > left_std_age_3) & (df.age < right_std_age_3)]
df_new_rating = df.loc[(df.player_rating > left_rating3) & (df.player_rating < right_rating3)]
df_new_rating
```

- **plot the histograms**

```python
# AGE DISTRIBUTION
plt.figure(figsize=(10, 6))

sns.histplot(data=df_new_age, x="age", kde=True)

# Mean
plt.axvline(mean_age, color="black", linestyle="-", linewidth=2, label="Mean")

# Standard Deviations
plt.axvline(left_std_age_1, linestyle="--", color="blue", label="-1 SD")
plt.axvline(right_std_age_1, linestyle="--", color="blue", label="+1 SD")

plt.axvline(left_std_age_2, linestyle="--", color="red", label="-2 SD")
plt.axvline(right_std_age_2, linestyle="--", color="red", label="+2 SD")

plt.axvline(left_std_age_3, linestyle="--", color="orange", label="-3 SD")
plt.axvline(right_std_age_3, linestyle="--", color="orange", label="+3 SD")

plt.title("Age Distribution with KDE")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.legend()

plt.show()
```
