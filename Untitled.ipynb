```python
days = [1,2,3,4,5]
temperature = [30,31,32,33,35]
humidity=[65,60,62,58,55]
```

```python
import matplotlib.pyplot as plt

plt.plot(days, temperature, label="Temperature")
plt.plot(temperature, humidity, label="Humidity")

plt.title("Temperature Conditions")
plt.xlabel("X-Label")
plt.ylabel("Y-Label")

plt.legend()
plt.grid(True)

plt.show()
```

```python
import matplotlib.pyplot as plt

days = [1,2,3,4,5]
sales = [10,15,12,18,20]

plt.figure(figsize=(8,4))

plt.plot(days,
         sales,
         marker="o",
         linestyle="-",
         color="blue",
         label="Sales"
        )

plt.title("Weekly Sales")
plt.xlabel("Days")
plt.ylabel("Sales")

plt.legend()
plt.grid(True)
plt.show()
```

```python
x = [1,2,3,4,5]
y = [2,5,3,7,6]
```

```python
import numpy as np

data = np.genfromtxt(
    "student_marks.csv",
    delimiter=",",
    skip_header=1
)

print(data)
```

```python
print(data.shape)
```

```python
days = data[:,0]
temperature = data[:,1]
```

```python
print(days, temperature)
```

```python
print(temperature.mean())
```

```python
print(temperature.max())
```

```python
print(temperature.min())
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,4))
plt.plot(days,
         temperature,
         marker="0",
         linestyle="-",
         color="blue",
         label="Climate"
        )

plt.title("Climate Conditions")
plt.xlabel("Days")
plt.ylabel("Temperature")

plt.lengend()
plt.grid(True)

plt.show()
```
