```python
import pandas as pd

df_sales = pd.read_excel("linechart.xlsx")
df_sales.head()
```

```python
# Line Chart
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 4))
plt.plot(df_sales["Quarter"], df_sales["Fridge"], color="blue", label="Fridge")
plt.plot(df_sales["Quarter"], df_sales["Dishwasher"], color="green", label="Dishwasher")
plt.plot(df_sales["Quarter"], df_sales["Washing Machine"], color="red", label="Washing Machine")

plt.title("Product Sales")
plt.ylabel("Revenue (Mill $)")
plt.xlabel("Financail Quarter")

plt.legend()

plt.show()
```

```python
total_sales = df_sales[["Fridge", "Dishwasher", "Washing Machine"]].sum()
```

```python
total_sales
```

```python
total_sales.index
```

```python
plt.pie(total_sales, labels=total_sales.index, autopct='%1.2f%%', shadow=True, explode=(0.1, 0, 0), startangle=120)

plt.show()
```

```python
#Bar Chart
df_sales.plot(kind="bar", x="Quarter")
plt.xticks(rotation=45)

plt.xlabel("Quarter")
plt.ylabel("Revenue")
plt.show()
```
